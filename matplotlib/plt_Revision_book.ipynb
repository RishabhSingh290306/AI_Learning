```python
import matplotlib.pyplot as plt
```

## Line plot / Line Chart

```python
x=[1,2,3,4]
y=[10,20,15,25]

plt.plot(x,y)
plt.show()
```

```python
x = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri']
y = [10,15,7,20,12]

plt.plot(x,y)
plt.xlabel('Days')
plt.ylabel('Sales')
plt.title("Bakery sales of the week")
plt.show()
```

```python
months = [1,2,3,4]
sales = [1000,1500,1200,1800]

plt.plot(months, sales, color ='blue', linestyle='--', linewidth =2, marker='o', label='2025 sales data')
plt.xlabel('Months')
plt.ylabel('Sales')

plt.legend(loc='upper left', fontsize = 12)
plt.grid(color ='gray', linestyle=':', linewidth=1)

plt.xlim(1,4) # x axis range from 1 to 4
plt.ylim(0,2000) # y axis range from 0 to 2000

plt.xticks([1,2,3,4], ['Jan', 'Feb', 'Mar', 'Apr']) # set x ticks and labels
plt.show()
```

## Bar charts

```python
import matplotlib.pyplot as plt

product=['Flour','Bread','Cake','Cookies']
sales = [1000, 1500, 800, 1200]

plt.bar(product, sales, color ='orange', label='Sales 2025')
plt.xlabel("product")
plt.ylabel('Sales')
plt.title('Product Sales')
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

product=['Flour','Bread','Cake','Cookies']
sales = [1000, 1500, 800, 1200]

plt.barh(product, sales, color ='orange', label='Sales 2025') # Horizontal bar chart

plt.xlabel("product")
plt.ylabel('Sales')
plt.title('Product Sales')
plt.legend()
plt.show()
```

## Pie Charts

```python
import matplotlib.pyplot as plt

regions = ['North','South','East','West']
sales =[100, 300, 400, 250]

plt.pie(sales, labels=regions, autopct='%1.1f%%', colors =['gold','skyblue','lightgreen','coral']) #autopct is used to show the percentage of each region in the pie chart
plt.title('Sales by Region')
```

## Histograms

```python
import matplotlib.pyplot as plt

scores = [45, 67, 89, 32, 100, 90, 78, 89, 90, 87, 98, 76]

plt.hist(scores, bins=5, color='skyblue', edgecolor='black') #bins is used to specify the number of bars in the histogram
plt.xlabel('Score')
plt.ylabel('Frequency')
plt.title('Histogram of Scores')
plt.show()
```

## ScatterPlot

```python
import matplotlib.pyplot as plt

hours_studies = [2, 3, 4, 5, 6, 7, 8, 9, 10]
exam_scores = [50, 60, 70, 75, 78, 80, 90, 95, 100]

plt.scatter(hours_studies, exam_scores, color='green', marker='^', label= 'Student data') #marker is used to specify the shape of the data points
plt.legend(loc='upper right')
plt.xlabel('Hours Studied')
plt.ylabel('Exam Score')
plt.grid()
plt.show()

```

```python
import matplotlib.pyplot as plt

plt.scatter([1,2,3],[50,55,60], color ='blue', label='Class A')
plt.scatter([1,2,3],[70,80,90], color='orange', label='Class B')

plt.legend(loc='upper left')
plt.xlabel('Hours Studied')
plt.ylabel('Exam Score')
plt.title("Comparing Exam Scores")
plt.grid()
plt.show()
```

## Essential Subplot Functions in Matplotlib

```python
import matplotlib.pyplot as plt

x=[1,2,3,4]
y=[50,60,70,80]

plt.subplot(1,2,1) # 1 row and 2 columns with the first plot
plt.plot(x,y)
plt.title("Line chart")

plt.subplot(1,2,2) # 1 row and 2 columns with the second plot
plt.bar(x,y)
plt.title("Bar Chart")

plt.tight_layout()
```

```python
import matplotlib.pyplot as plt

x =[1,2,3,4]
y=[50,60,70,80]

fig, ax = plt.subplots(1, 2, figsize =(10,5))

ax[0].plot(x,y)
ax[0].set_title("Line Chart")

ax[1].bar(x,y, color='green')
ax[1].set_title("Bar Chart")

fig.suptitle("comparison of line and bar chart")

fig, ax = plt.subplots(2,1, sharex=True) # sharex means that the x-axis will be shared between both subplots

plt.tight_layout()
plt.show()

```

## Saving graphs as files

```python
import matplotlib.pyplot as plt

x=[1,2,3,4]
y=[10,20,15,25]

plt.plot(x,y, color='red', marker='o')
plt.title("Simple Line plot")
plt.xlabel('X-axis')
plt.ylabel('Y-axis')

plt.savefig('myplot.png', dpi=300, bbox_inches='tight')
plt.show()
```
