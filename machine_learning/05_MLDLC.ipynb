# Machine Learning Development Life Cycle (MLDLC)

It is the complete process of building, deploying, and maintaining a machine learning system, from understanding the problem to monitoring the model in production.

Think of it as the ML equivalent of the Software Development Life Cycle (SDLC).

## 1. Frame the problem
Define: What problem are we solving? What is the expected outcome? How will success be measured?

Example:
A telecom company wants to predict customer churn.

Goal: Identify customers likely to leave.

## 2. Gathering Data

Gather relevant data from:

1. Databases
2. APIs
3. Sensors
4. User interactions
5. Public datasets

Example:
For churn prediction:

1. Customer age
2. Monthly bill
3. Contract type
4. Usage history

We do not prefer modelling our data on dynamic data or while web scraping from the internet.
Since false or malicious data can break our model, so we build data warehouses where we store static data and use it for training our models.


## 3. Data Preprocessing

Clean and transform the data.

Tasks include:

1. Handling missing values
2. Removing duplicates
3. Encoding categorical variables
4. Feature scaling
5. Feature engineering

Example:
Convert:

==> "Male", "Female" → 0, 1\
==> Fill missing ages with median age


## 4. Exploratory Data Analysis (EDA)
Understand the data using statistics and visualizations.

Questions:\
Which features are important?\
Are there outliers?\
Are classes balanced?

Tools:\
Pandas\
Matplotlib\
Seaborn




## 5. Feature Engineering and Selection
Create useful features and remove irrelevant ones.

Example:\
Instead of: 

Purchase Date\
Last Visit Date

Create:

Days Since Last Purchase

This often improves model performance.


## 6. Model Training, Evaluation and Selection

Choose an appropriate algorithm.\
Train the model using training data. The model learns patterns by minimizing error.\
Check how well the model performs on unseen data.

### Hyperparameter Tuning
Hyperparameters (Set by Humans) are settings that control how the model learns.

|Algorithm|	        Hyperparameter|
|---|---|
|KNN	            |Number of neighbors (K)|
|Random Forest	    |Number of trees|
|Decision Tree	    |Maximum depth|
|Neural Network	    |Learning rate, batch size, epochs|

You choose these before training.

Why Tune Hyperparameters?\
Different values can produce very different results.

For example, in KNN:

K = 1 → May overfit\
K = 50 → May underfit\
K = 7 → Might work best

The goal is to find the values that give the best performance on unseen data.

### What is Ensemble Learning?

It is a tuning technique in which instead of relying on one model, combine multiple models and let them make a joint decision.

Idea:

Many weak learners together can become a strong learner.

## 8. Testing
Testing means evaluating the trained model on data it has never seen before.

Why do we test?

A model might perform very well on training data simply because it memorized it. Testing tells us whether it can generalize to new examples.

## 9. Optimization

Optimization means improving the model's performance after testing reveals weaknesses.

Methods of Optimization\
A. Hyperparameter Tuning\
B. Feature Engineering\
C. Feature Selection\
D. Collect More Data\
E. Use a Better Algorithm\
F. Ensemble Methods
