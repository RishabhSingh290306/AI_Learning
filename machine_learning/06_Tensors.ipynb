# Tensors

A tensor is simply a way of representing data in multiple dimensions.

You can think of tensors as a generalization of:

|Object	|Dimensions|
|---|---|
|Scalar |(single number)	|0D|
|Vector |(list of numbers)|	1D|
|Matrix |(table of numbers)|	2D|
|Tensor|	3D, 4D, 5D, ...|

```python
import numpy as np
```

```python
a = np.array(4)
a  
#This is 0D tensor
```

```python
#You can find the dimension 
# Number of axis (or rank) = dimension of the tensor
a.ndim
```

```python
# A list of numbers(vector or 1D array) like [1,2,3,4] is a 1 D tensor.

arr = np.array([1,2,3,4])
arr
```

```python
arr.ndim
```

A very confusing thing is that a 1D Tensor is a vector. So arr here is a 1D tensor or vector.
When we talk about the dimension of the vector we say that it is 4 (the number of elements in the vector). Also, notice when we combine the scalar 1,2,3 and 4 we get a vector.
