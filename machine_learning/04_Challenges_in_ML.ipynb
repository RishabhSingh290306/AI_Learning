# Challenges in Machine Learning

## Data Collection

Collecting data for an enterprise is a complex task that requires careful planning and execution.
we can either use APIs or web-scraping.

## 1. Insufficient data/labelled data
Machine learning models learn patterns from examples. If you have too little data, the model may fail to generalize.

Example

Imagine trying to identify cats and dogs using only 20 images. The model won't see enough variation to learn reliable patterns.

Problem: Underfitting and poor performance on new data.

## 2. Non-Representative Data 
The training data should represent the real-world situations where the model will be used.

Example

Suppose a weather prediction model is trained only on summer data.

When winter arrives, its predictions may be poor because it has never seen winter conditions.




## 3. Sampling Bias
Data may unintentionally favor certain groups.

Example

A salary prediction model trained mostly on data from one city may perform poorly for people in other cities.

## 4. Poor Data Quality
Collected data may contain:

Missing values
Incorrect entries
Duplicate records
Noise

## 5. Irrelevant features
An irrelevant feature is a feature that has little or no relationship with the target variable.

Example

Suppose you're predicting house prices.

Area	Bedrooms	Distance from City	Owner's Favorite Color	Price
1200	2	5 km	Blue	₹25 lakh

Here:

Area ✅ Relevant
Bedrooms ✅ Relevant
Distance from City ✅ Relevant
Owner's Favorite Color ❌ Irrelevant


## 6. Overfitting
Overfitting occurs when a model learns the training data too well, including noise and random patterns.

Example

Imagine memorizing previous exam questions instead of understanding concepts.

You score well on familiar questions but fail on new ones.

Visualization

Training Accuracy = 99%

Test Accuracy = 65%

This is a sign of overfitting.

Characteristics

✅ Excellent performance on training data

❌ Poor performance on unseen data

## 7. Underfitting
Underfitting occurs when a model is too simple to learn the underlying pattern.

Example

Trying to predict house prices using only:

Price = Constant

Clearly too simple.

Visualization

Training Accuracy = 60%

Test Accuracy = 55%

Poor performance everywhere.

Characteristics

❌ Poor training performance

❌ Poor testing performance


## 8. Software Integration
Building a model is only part of the work.

The model must be integrated into a real application.

Example

A recommendation model for an e-commerce website must work with:

Frontend,
Backend,
Databases,
APIs,
User authentication systems,
Challenges,
Compatibility issues,
Different technologies,
Real-time requirements,
Scalability concerns

In real world, many ML projects fail not because the model is bad, but because integrating it into existing systems is difficult.

## 9. Offline learning/Deployment
In Offline (Batch) Learning, the model is trained once and then deployed.

Workflow

Training Data → Train Model → Deploy Model → Predict

The deployed model does not learn automatically from new data.

Example

A spam detection model trained in January.

By June, spammers may use new tricks.

The model still relies on January's knowledge.

Challenges : Performance degrades over time ,Requires periodic retraining, Updating models can be expensive

Solution : Regular retraining with fresh data
