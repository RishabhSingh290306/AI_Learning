# Batch(Offline) V/S Online ML

## 1. Batch Learning (Offline Learning)

In batch learning, the model is trained using the entire dataset at once.

--> you collect all training data 
--> train the model on an offline system 
--> deploy the model on the server
--> if new data arrives, retrain the model from scratch (or with the old + new data)

The problem with batch learning :
The model doesn't learn automatically. You need to retrain it.
--> lots of data
--> hardware limitation 
--> availability 

## 2. Online Learning 

In online learning, the model learns continuously as new data arrives.

--> Train on a small amount of data.
--> New data arrives.
--> Update the model immediately.
--> Repeat forever.

when to use?
1. where there is concept drift
2. cost effective
3. faster solution

There are many useful libraries for online machine learning to work on live data like river and vowpal wabbit.

### Learning Rate
 The frequency at which we retrain the model on new data. 
 Setting the right learning rate for your model is very important since higher learning rate may cause the model to retrain frequently and forget old data and lower learning rate may cause the model to learn slowly 

### Out of core learning
 A learning technique used when the training dataset is too large. So you divide the dataset into small packages and send it to the server, so it falls under the online learning category.

### Disadvantages of Online learning
 1. tricky to use
 2. sensistive to bad data
 3. can forget old patterns
 4. needs continuos monitoring
 5. vulnerable to data drift
