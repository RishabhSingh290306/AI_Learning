# Univariate Analysis
 the independent analysis of a single column(variable)

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("assets/data/train.csv")
```

```python
df.head()
```

First we will work on Categorical Data


## 1. Categorical Data

### a. countplot

```python
# lets work on survived and passenger class i.e pclass
sns.countplot(x='Survived',data = df)

df['Survived'].value_counts()
```

```python
sns.countplot(x="Pclass", data=df)
df['Pclass'].value_counts()
```

```python

df["Embarked"].value_counts().plot(kind='bar')
```

### b. Piechart

```python
df['Survived'].value_counts().plot(kind='pie',autopct='%.2f')
```

## 2. Numerical Data


### a. Histogram


```python
plt.hist(df['Age'],bins=10)
```

### b. Displot

```python
sns.displot(df['Age'], kde= True) 
```

kde is kernel density estimation. It's a smooth curve that estimates the probability density of a continuous variable. Think of it as a smoothed version of a histogram.

### c. Boxplot

```python
sns.boxplot(x='Age',data=df)
```

```python
df['Age'].mean()
```

```python
df['Age'].min()
```

```python
df['Age'].max()
```

```python
df['Age'].skew()
```

if skewness is 0 then its perfectly skewed in the middle\
if its negative , its skewed left side\
