# Instance V/S Model Based Learning

## 1. Instance Based Learning (Lazy learners)

Instance-based learning memorizes the training examples and compares new data points to those stored examples when making predictions. There is no learning here.

Think of it as:

"I don't learn a formula. I just remember past cases and find the most similar one."

Example

Suppose you're learning to identify fruits.

Training Data:

Weight	Color	Fruit
150g	Red  	Apple
170g	Red 	Apple
120g	Orange	Orange

Now a new fruit arrives:

Weight = 160g
Color = Red

The algorithm looks at stored examples and says:

"This looks very similar to those red fruits weighing around 150–170g, so it's probably an Apple."

Popular Instance-Based Algorithms
--> K-Nearest Neighbors (KNN)
--> Case-Based Reasoning

### Disadvantages

❌ Requires storing all training data
❌ Prediction can be slow
❌ Doesn't scale well for huge datasets

## 2. Model Based learning ( conventional learning)

Model-based learning tries to learn a general rule or formula from the data.

Think of it as:

"Instead of memorizing examples, I'll discover the underlying pattern."

Example

Suppose you're predicting house prices.

Training Data:

Area (sq ft)	Price
1000	        ₹20 lakh
1500	        ₹30 lakh
2000	        ₹40 lakh

The algorithm learns a relationship like:

Price=20,000×Area

Now, for a 1800 sq ft house, it calculates the price using the learned model.

Popular Model-Based Algorithms
--> Linear Regression
--> Logistic Regression
--> Decision Trees
--> Random Forests
--> Neural Networks

### Disadvantages

❌ Training can take time
❌ Risk of underfitting or overfitting
❌ Choosing the right model can be difficult
