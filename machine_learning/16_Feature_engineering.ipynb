# Feature Engineering

Feature engineering is the process of using domain knowledge to extract features from raw data.\
These features can be used to improve the performance of machine learning algorithms

1. Missing Values imputation\
Missing values occur when some data is unavailable.\
Machine learning models usually cannot handle NaN values.\
Replace missing values with the average.



2. Handling Categorical Variables\
Machine learning models work with numbers, not text.\
Need conversion into numerical format.(One hot Encoding)


3. Outlier Detection\
Outliers are unusually large or small values.\
Values outside the Inter Quartile Range are outliers.



4. Feature Scaling\
Different features can have different ranges.\
Scaling puts features on a similar scale.


5. Feature Construction\
Creating new features from existing ones.\
Often improves model performance significantly.


6. Feature Selection\
Choosing only the most useful features.\
Why? Not all columns help prediction.

7. Feature Extraction\
Converts many features into fewer informative features.\
Unlike feature selection, it creates new features.
