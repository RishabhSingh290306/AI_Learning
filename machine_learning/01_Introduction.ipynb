# Introduction

## What is machine learning?

Machine learning is a field of CS that uses statistical techniques to give computer systems the ability to "Learn" with data without being explicitly programmed.

## Where ML becomes useful?

scenario 1 : You want to build an email classifier that identifies spam from non spam emails

scenario 2 : Image classification for dog breed , you cannot write explicitly all the cases using if else

scenario 3 : Data mining ( where you apply a ML model on the data to find patterns and make predictions)

## Difference Between AI ML and DL

you never explicitly code in ML, which is the case in symbolic AI.
AI fails in case of fuzzy logic.
Deep learning has the same flow as ML.
The class of algorithms in deep learning are inspired by our Biology. The perceptron in DL is inspired by neuron.
Deep learning is very useful in Fuzzy logic problems.
Accuracy and performance of deep learning increase with data input.
