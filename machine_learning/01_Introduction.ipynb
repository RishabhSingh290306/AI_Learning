## Types of ML (based on the amount of supervision need)

1. Supervised   2. Unsupervised    3. Semisupervised    4. Reinforcement

## 1. Supervised Learning

Supervised Learning is a type of Machine Learning where the model learns from labeled data.

Think of it like learning with an answer key.

You provide the algorithm with:

Input (X) → features/data
Output (Y) → correct answer/label

The model learns the relationship between X and Y so it can predict Y for new unseen data.

## What is regression and classification?

Regression is a type of supervised learning used when the target/output is a continuous numerical value.

Classification is a type of supervised learning where the goal is to predict a category or class label instead of a numerical value.

## 2. Unsupervised Learning

Unsupervised Learning is a type of Machine Learning where the model is given data without labels and must discover patterns, structures, or relationships on its own.

## types of Unsupervised learning

a. Clustering 

Group similar data points together.
The algorithm tries to answer: "Which data points look alike?" without being told the correct groups.

b. Dimensionality reduction

Reduce the number of features while preserving important information.
Imagine a student dataset:

Name	Age	Height	Weight	BMI	Marks

Many features may contain overlapping information.
For example:
Height and Weight contribute to BMI. Some features may be redundant.

c. Anomaly Detection

Find unusual or abnormal data points.
The algorithm tries to answer: "Which points don't look like the rest?"

d. Association Rule Learning

Discover relationships between items.
The algorithm asks: "What items tend to occur together?"

## 3. Semisupervised Learning

Semi-Supervised Learning is a machine learning approach that uses:
A small amount of labeled data ✅
A large amount of unlabeled data ✅
to train a model. It's a combination of Supervised Learning and Unsupervised Learning.

Why do we need it? Labeling data is expensive and time-consuming.

## 4. Reinforcement learning

Reinforcement Learning (RL) is a type of Machine Learning where an agent learns by interacting with an environment and receiving rewards or penalties for its actions.

The goal is simple:
Maximize total reward over time.

AlphaGo defeated world champions in Go using RL.
