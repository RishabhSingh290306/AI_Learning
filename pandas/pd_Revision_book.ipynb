```python
df['SepalLengthCm'] = df['SepalLengthCm'].astype(float)  # Convert to float
```

```python
df1 = pd.DataFrame({
    'Name': ['David', 'Eve'],
    'Age': [28, 22],
    'City': ['Miami', 'Seattle']
})
print(df1)
```

```python
df2 = pd.DataFrame({
    'Name': ['Frank', 'Grace'], 
    'Age': [33, 27],
    'City': ['Austin', 'Denver']
})
print(df2)
```

```python
df3 = pd.concat([df1, df2], ignore_index=True)  # Concatenate DataFrames and reset index
print(df3)
```

```python
d4 = pd.merge(df1, df2, on='Name', how='inner')  # Merge DataFrames on 'Name' column using inner join
print(d4)
```
