```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('D:\Projects\AI_Learning\seaborn\Data\employees.csv', encoding='latin1')
df.head()
```

```python
sns.set(style ='whitegrid')
```

```python
sns.histplot(df['SALARY'], kde = True, bins =10)
plt.title("Distribution of Salaries")
```

```python
sns.boxplot(x ='DEPARTMENT_ID', y ='SALARY', data =df)
plt.title("Salary Distribution by Department")
```

```python
sns.violinplot(x ='DEPARTMENT_ID', y ='SALARY',hue = 'DEPARTMENT_ID', data =df, split =True)
plt.title("Age distribution by department")
```

```python
sns.scatterplot(x='MANAGER_ID',y='SALARY',data=df,hue='DEPARTMENT_ID', palette ='bright')
plt.title('Salary Distribution by Manager')
```

```python
sns.pairplot(df[['EMPLOYEE_ID','SALARY']], kind='reg')
plt.suptitle("Pair Plot of Employee Salary and Hire Date",y=1.02)
```

```python
corr = df[["EMPLOYEE_ID","SALARY"]].corr()
sns.heatmap(corr, annot=True, cmap ='coolwarm')
plt.title('Correlation Matrix for Employee ID and Salary')
```

```python
sns.countplot(x ='DEPARTMENT_ID',hue= 'DEPARTMENT_ID', data=df)
plt.title(("Count of Employees by Department"))
```

```python
sns.regplot(x="EMPLOYEE_ID", y="SALARY",data = df)
plt.title("Regression between Employee ID and Salary")
```
