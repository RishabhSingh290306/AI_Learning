# CTEs (Common Table Expressions)

```python
"""""
First, what problem do CTEs solve?

Suppose you have a query like:

SELECT *
FROM (
    SELECT customer_id,
           SUM(amount) AS total_spent
    FROM orders
    GROUP BY customer_id
) t
WHERE total_spent > 10000;

This works.

But imagine the query becomes 100 lines long with multiple nested subqueries.

Then you get:

SELECT *
FROM (
    SELECT *
    FROM (
        SELECT ...
    ) a
) b;

It becomes difficult to read and debug.

CTEs solve this.
"""
```

```python
"""
Think of a CTE as:

"Create a temporary named table and use it immediately."

Just like creating a variable in programming.

Python:

total = a + b
print(total)

SQL CTE:

WITH sales_summary AS (
    SELECT customer_id,
           SUM(amount) AS total_sales
    FROM orders
    GROUP BY customer_id
)

SELECT *
FROM sales_summary;
Basic Syntax
WITH cte_name AS (
    query
)

SELECT *
FROM cte_name;

Read it as:

"WITH this temporary table available, run the next query."
"""
```

```python
"""
Multiple CTEs

You can chain them.

WITH customer_spending AS (
    SELECT customer_id,
           SUM(amount) AS total_spent
    FROM orders
    GROUP BY customer_id
),

high_value_customers AS (
    SELECT *
    FROM customer_spending
    WHERE total_spent > 1000
)

SELECT *
FROM high_value_customers;

Notice:

high_value_customers

is using

customer_spending

created just above it.

This is very common in interviews.
"""
```

```python
"""Key Lesson From Today's Class

When you see a complex problem, ask:

"What intermediate table do I need first?"

For this problem, the intermediate table was:

customer_spending

Once you had that, the rest became easy.

That's the core mindset behind CTEs."""
```
