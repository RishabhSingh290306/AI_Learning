# Set Operations

### 1.Union

```python
Suppose:

current_customers
customer_id
101
102
103
old_customers
customer_id
103
104
105

Query:

SELECT customer_id
FROM current_customers

UNION

SELECT customer_id
FROM old_customers;

Result:

customer_id
101
102
103
104
105

Notice:

103

appears only once.
```

```python
Important

UNION removes duplicates.

Think:

UNION = Combine + Deduplicate
```

### 2. Union All

```python
Same tables.

SELECT customer_id
FROM current_customers

UNION ALL

SELECT customer_id
FROM old_customers;

Result:

customer_id
101
102
103
103
104
105

Now duplicates remain.
```

```python
Interview Favorite

Which is faster?

UNION

or

UNION ALL

Answer:

UNION ALL

Why?

Because PostgreSQL doesn't have to remove duplicates.
```

### 3. INTERSECT

```python
Find common rows.

current_customers
customer_id
101
102
103
old_customers
customer_id
103
104
105

Query:

SELECT customer_id
FROM current_customers

INTERSECT

SELECT customer_id
FROM old_customers;

Result:

customer_id
103

Only common values.
```

### 4. EXCEPT

```python
Find values in first query but not second.

Query:

SELECT customer_id
FROM current_customers

EXCEPT

SELECT customer_id
FROM old_customers;

Result:

customer_id
101
102

Because:

103

exists in both tables.
```

### Rule for Set Operations


Both queries must return:

Same number of columns

Valid:

SELECT customer_id
UNION
SELECT customer_id

Invalid:

SELECT customer_id, name
UNION
SELECT customer_id
Compatible Data Types

Valid:

INT
UNION
INT

Invalid:

INT
UNION
DATE

(PostgreSQL will complain.)
