# SQL Analytics Case Studies

These are the 7 core business analytics case studies every Data Analyst should know. They combine SQL concepts with real-world business problems.

### 1. Customer Retention Analysis

```python
Business Question

How many customers return and make additional purchases after their first purchase?

Objective

Identify:

New Customers
Returning Customers
SQL Concepts Used
CTEs
MIN()
JOIN
CASE
Analyst Thinking
Step 1

Find each customer's first purchase date.

MIN(order_date)
Step 2

Join back to all orders.

Step 3

Classify:

CASE
    WHEN order_date = first_purchase_date
    THEN 'New'
    ELSE 'Returning'
END


Business Insight : 
1. Customer acquisition
2. Customer retention
3. Repeat purchase behavior
```

### 3. Top Customers Per Region

```python
Business Question

Who are the highest-spending customers in each region?

SQL Concepts Used
1. JOIN
2. SUM()
3. GROUP BY
4. ROW_NUMBER()
5. Analyst Thinking

Step 1

Calculate customer spending.

Step 2

Join customer region.

Step 3

Rank customers within each region.

ROW_NUMBER() OVER (
    PARTITION BY region
    ORDER BY total_spent DESC
)

Step 4

Keep rank = 1.

Business Insight :
1. Regional champions
2. Territory analysis
3. Sales performance
```

### 4. Revenue Analytics

```python
Business Question

How is revenue changing over time?

SQL Concepts Used
1. DATE_TRUNC()
2. SUM()
3. LAG()
4. Window Functions
5. Running Totals

KPIs
Monthly Revenue
SUM(amount)

Previous Month Revenue
LAG(revenue)

Revenue Growth
revenue - previous_revenue

Growth %
(Current - Previous)
/
Previous
*100

Cumulative Revenue
SUM(revenue)
OVER(ORDER BY month)

Business Insight :
1. Revenue trends
2. Growth tracking
3. Executive reporting
```
