# Recursive CTEs

```python
You already know:

WITH customer_spending AS (
    ...
)
SELECT *
FROM customer_spending;

That's a normal CTE.

A Recursive CTE is a CTE that refers to itself.

Think:

"Keep repeating this query until a condition is met."
```

### Why Do We Need It?

```python
Imagine a company hierarchy.
| emp_id | employee | manager_id |
| ------ | -------- | ---------- |
| 1      | CEO      | NULL       |
| 2      | Alice    | 1          |
| 3      | Bob      | 1          |
| 4      | Charlie  | 2          |
| 5      | David    | 2          |
| 6      | Emma     | 4          |


Question:

Who works under the CEO?

Not just direct reports.

We want:

CEO
├── Alice
│   ├── Charlie
│   │   └── Emma
│   └── David
└── Bob

This is a hierarchy.
```

### Structure of a Recursive CTE

```python
A recursive CTE has 2 parts:

1. Anchor Query

Starting point.

2. Recursive Query

Keeps finding more rows.

Skeleton
WITH RECURSIVE cte_name AS (

    -- Anchor Query

    SELECT ...

    UNION ALL

    -- Recursive Query

    SELECT ...
    FROM table t
    JOIN cte_name c
      ON ...

)

SELECT *
FROM cte_name;
```

### Simple Example: Counting

```python
Suppose we want:

1
2
3
4
5

without a table.

Query
WITH RECURSIVE numbers AS (

    SELECT 1 AS n

    UNION ALL

    SELECT n + 1
    FROM numbers
    WHERE n < 5

)

SELECT *
FROM numbers;


How It Runs
Step 1 (Anchor)
1
Step 2
1 + 1 = 2
Step 3
2 + 1 = 3
Step 4
3 + 1 = 4
Step 5
4 + 1 = 5

Stop

Because:

WHERE n < 5

becomes false.

Result
n
1
2
3
4
5
```

### Interview Favorite: Employee Hierarchy

```python
Table:
| emp_id | employee | manager_id |
| ------ | -------- | ---------- |
| 1      | CEO      | NULL       |
| 2      | Alice    | 1          |
| 3      | Bob      | 1          |
| 4      | Charlie  | 2          |

Query:

WITH RECURSIVE employee_tree AS (

    SELECT
        emp_id,
        employee,
        manager_id
    FROM employees
    WHERE manager_id IS NULL

    UNION ALL

    SELECT
        e.emp_id,
        e.employee,
        e.manager_id
    FROM employees e
    JOIN employee_tree et
      ON e.manager_id = et.emp_id

)

SELECT *
FROM employee_tree;
What Happens?

Start with:

CEO

Then find:

Employees reporting to CEO

Then:

Employees reporting to them

Then:

Employees reporting to them

until none remain.

Mental Model

Think of recursion like:

def explore(node):
    for child in node.children:
        explore(child)

Recursive CTEs bring that idea into SQL.
```
