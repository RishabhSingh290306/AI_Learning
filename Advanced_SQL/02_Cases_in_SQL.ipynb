# CASE Statements

```python
"""Why CASE Exists

Imagine your manager asks:

"Can you label customers based on spending?"

SQL can't magically create labels. We need logic like:

IF total_spent > 5000
    Premium

ELSE IF total_spent > 1000
    Regular

ELSE
    Basic

In Python, you'd write:

if total_spent > 5000:
    customer_type = "Premium"
elif total_spent > 1000:
    customer_type = "Regular"
else:
    customer_type = "Basic"

In PostgreSQL, the equivalent is CASE."""
```

```python
"""CASE Syntax
CASE
    WHEN condition1 THEN result1
    WHEN condition2 THEN result2
    ELSE result3
END

Think of it as SQL's version of if-elif-else.

Example
SELECT
    customer_id,
    total_spent,
    CASE
        WHEN total_spent > 5000 THEN 'Premium'
        WHEN total_spent > 1000 THEN 'Regular'
        ELSE 'Basic'
    END AS customer_type
FROM customer_spending;"""
```

```python
"""
| customer_id | total_spent | customer_type |
| ----------- | ----------- | ------------- |
| 101         | 7000        | Premium       |
| 102         | 3000        | Regular       |
| 103         | 500         | Basic         |
"""
```

```python
"""
Very Important Interview Concept

Order matters.

Suppose you write:

CASE
    WHEN total_spent > 1000 THEN 'Regular'
    WHEN total_spent > 5000 THEN 'Premium'
    ELSE 'Basic'
END

For a customer with:

total_spent = 7000

The first condition is already true:

7000 > 1000

So SQL stops there and returns:

Regular

which is wrong.

Therefore:

WHEN total_spent > 5000

must come before

WHEN total_spent > 1000"""
```
