# Query Optimization

### SELECT *

```python
Suppose a table has:

customer_id
name
email
phone
address
city
state
country
...
20 more columns

You need only:

customer_id
name

Bad:

SELECT *
FROM customers;

Better:

SELECT
    customer_id,
    name
FROM customers;

Why?

Because PostgreSQL reads less data.
```

### Filtering Early

```python
Bad:

SELECT *
FROM orders
WHERE amount > 10000;

after joining several huge tables.

Better:

SELECT *
FROM orders
WHERE amount > 10000

before joining if possible.

Think:

Reduce data first, then do expensive work.
```

### Avoid Functions in WHERE

```python
Suppose:

SELECT *
FROM orders
WHERE EXTRACT(YEAR FROM order_date) = 2025;

Works.

But often prevents efficient index usage.

Better:

SELECT *
FROM orders
WHERE order_date >= '2025-01-01'
  AND order_date < '2026-01-01';

Same result.

Usually better performance.

This is a favorite interview topic.
```

### EXISTS vs IN

```python
Suppose:

SELECT *
FROM customers
WHERE customer_id IN (
    SELECT customer_id
    FROM orders
);

Often fine.

But for large datasets:

SELECT *
FROM customers c
WHERE EXISTS (
    SELECT 1
    FROM orders o
    WHERE o.customer_id = c.customer_id
);

can perform better.

Don't memorize this as a rule—always check with EXPLAIN ANALYZE.
```

### JOIN vs Subquery

```python
Beginners often write deeply nested subqueries.

Example:

SELECT *
FROM (
    SELECT *
    FROM (
        SELECT *
        FROM orders
    ) t
) x;

Hard to read.

Usually a CTE or join is cleaner.

Remember:

WITH customer_spending AS (...)

is often easier to understand and maintain.
```

### The 80/20 Rule

```python
For Data Analysts, 80% of performance issues come from:

Using SELECT *
Missing indexes
Poor filtering
Unnecessary joins
Not understanding execution plans

You don't need to become a PostgreSQL performance engineer.

You just need to avoid the biggest mistakes.
```
