# Running Totals

```python
The Problem

Table:
| month | revenue |
| ----- | ------- |
| Jan   | 100     |
| Feb   | 200     |
| Mar   | 300     |
| Apr   | 400     |

Management asks:

Show cumulative revenue.

Normal SUM()
SELECT SUM(revenue)
FROM sales;

Result:

1000

Problem:

We only get one number.

What We Want
| month | revenue | running_total |
| ----- | ------- | ------------- |
| Jan   | 100     | 100           |
| Feb   | 200     | 300           |
| Mar   | 300     | 600           |
| Apr   | 400     | 1000          |

Notice:
Jan = 100
Feb = 100 + 200
Mar = 100 + 200 + 300
Apr = 100 + 200 + 300 + 400
```

```python
The Solution

SELECT
    month,
    revenue,
    SUM(revenue) OVER (
        ORDER BY month
    ) AS running_total
FROM sales;

How It Thinks

For January:
100
For February:
100 + 200
For March:
100 + 200 + 300
For April:
100 + 200 + 300 + 400
```

```python
Visualizing the Window

Think:

SUM(revenue) OVER (
    ORDER BY month
)

means:

For each row:

Sum everything from the beginning up to the current row.



Real Analyst Example

Monthly sales:
| month | revenue |
| ----- | ------- |
| Jan   | 5000    |
| Feb   | 7000    |
| Mar   | 6000    |
| Apr   | 9000    |

Query:

SELECT
    month,
    revenue,
    SUM(revenue) OVER (
        ORDER BY month
    ) AS cumulative_revenue
FROM sales;

Result:
| month | revenue | cumulative_revenue |
| ----- | ------- | ------------------ |
| Jan   | 5000    | 5000               |
| Feb   | 7000    | 12000              |
| Mar   | 6000    | 18000              |
| Apr   | 9000    | 27000              |
```

```python
Partitioned Running Totals

Suppose:
| department | month | revenue |
| ---------- | ----- | ------- |
| IT         | Jan   | 100     |
| IT         | Feb   | 200     |
| HR         | Jan   | 50      |
| HR         | Feb   | 80      |

We want cumulative revenue within each department.

SELECT
    department,
    month,
    revenue,
    SUM(revenue) OVER (
        PARTITION BY department
        ORDER BY month
    ) AS running_total
FROM sales;

Result:

IT
100
300
HR
50
130

The running total restarts because of:

PARTITION BY department
```

```python
Important PostgreSQL Concept: Window Frames

What PostgreSQL actually uses is:

SUM(revenue) OVER (
    ORDER BY month
    ROWS BETWEEN UNBOUNDED PRECEDING
    AND CURRENT ROW
)

Looks scary.

But it simply means:

Start at the first row and keep summing until the current row.

This is the default behavior when you write:

SUM(revenue) OVER (
    ORDER BY month
)

So don't worry about the long version yet.
```
