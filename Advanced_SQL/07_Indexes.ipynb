## Indexes

```python
Module 13: Indexes
Imagine This Situation

You have a table:

customers

with:

10,000,000 rows

You run:

SELECT *
FROM customers
WHERE customer_id = 987654;

How does PostgreSQL find that customer?
```

```python
Without an Index

PostgreSQL does:

Check row 1
Check row 2
Check row 3
...
Check row 9,876,54

This is called a:

Sequential Scan

Think:

Looking for a name in a notebook by reading every page.

Slow.
```

```python
With an Index

Suppose:

CREATE INDEX idx_customer_id
ON customers(customer_id);

Now PostgreSQL creates a special structure.

Think:

Book Index

At the back of a textbook:

Photosynthesis → Page 105
SQL → Page 312

You don't read the whole book.

You jump directly to the page.
```

```python
Mental Model

Without Index:

Search Every Row

With Index:

Jump Near The Answer
```

## PostgreSQL Default Index

```python
Most common:

CREATE INDEX idx_customer_id
ON customers(customer_id);

This creates a:

B-Tree Index

You don't need to know the internals yet.

Just remember:

PostgreSQL Default = B-Tree
```

## Explain

```python
This is one of the most useful commands in PostgreSQL.

Suppose:

SELECT *
FROM customers
WHERE customer_id = 101;

Write:

EXPLAIN
SELECT *
FROM customers
WHERE customer_id = 101;

PostgreSQL tells you:

How it plans to execute the query

Think:

"Show me your strategy before running."
```

## Explain Analyze 

```python
Even better:

EXPLAIN ANALYZE
SELECT *
FROM customers
WHERE customer_id = 101;

Now PostgreSQL actually runs the query and tells you:

What happened
Actual execution time
Rows scanned
Whether index was used

This is the foundation of query optimization.
```
