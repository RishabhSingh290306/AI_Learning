# Date Functions

```python

Most business tables contain dates:

| order_id | order_date |
| -------- | ---------- |
| 1        | 2025-01-15 |
| 2        | 2025-02-10 |


Analysts are constantly asked:

Revenue by month
Orders this year
Customers acquired last quarter
Growth month-over-month
Retention by signup month

To do that, you need date functions.
```

```python
Getting Current Date

SELECT CURRENT_DATE;

Example output:

2026-06-20
```

```python
Extracting Parts of a Date

Suppose:

order_date = '2025-03-15'
Year
SELECT EXTRACT(YEAR FROM order_date)
FROM orders;

Result:
2025

Month
SELECT EXTRACT(MONTH FROM order_date)
FROM orders;

Result:
3

Day
SELECT EXTRACT(DAY FROM order_date)
FROM orders;

Result:
15
```

```python
Another Very Common Function
DATE_TRUNC()

This is extremely important in PostgreSQL.

Suppose:

2025-03-15 14:37:12
Month Level
SELECT DATE_TRUNC('month', order_date)
FROM orders;

Result:

2025-03-01 00:00:00

All March dates become the first day of March.

This makes monthly reporting much easier.
```

```python
Most Common Analyst Query
Orders by Month

| order_id | order_date |
| -------- | ---------- |
| 1        | 2025-01-10 |
| 2        | 2025-01-15 |
| 3        | 2025-02-01 |

Query:

SELECT
    EXTRACT(MONTH FROM order_date) AS month,
    COUNT(*) AS total_orders
FROM orders
GROUP BY month
ORDER BY month;

Result:
| month | total_orders |
| ----- | ------------ |
| 1     | 2            |
| 2     | 1            |
```
