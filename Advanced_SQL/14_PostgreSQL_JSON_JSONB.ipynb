# PostgreSQL JSON & JSONB

```python
One reason PostgreSQL is so popular is that it can store both:

Traditional SQL Data
customer_id | name | age

and

JSON Data
{
  "name": "Rishabh",
  "age": 20,
  "city": "Bengaluru"
}

inside the same database.
```

### Why JSON Exists

```python
Suppose you're storing customer preferences.

Traditional SQL:

customer_id | language | theme | notifications

Works.

But what if every customer has different settings?

Customer A:

{
  "theme": "dark"
}

Customer B:

{
  "theme": "light",
  "language": "english",
  "notifications": true
}

JSON handles flexible data very well.
```

### JSON vs JSONB

```python
PostgreSQL provides:

JSON

Stores text exactly as provided.

JSONB

Stores a binary optimized version.

In practice:

Use JSONB

about 95% of the time.

Example Table
CREATE TABLE customers (
    customer_id INT,
    profile JSONB
);

Data:

{
  "name": "Rishabh",
  "city": "Bengaluru",
  "age": 20
}
```

### Extracting Values

```python
Operator: ->

Returns JSON.

SELECT profile -> 'name'
FROM customers;

Result:

"Rishabh"

Notice the quotes.

Still JSON.

Operator: ->>

Returns text.

SELECT profile ->> 'name'
FROM customers;

Result:

Rishabh

No quotes.

Text value.
```

### Filtering JSON Data

```python
Suppose:

{
  "name": "Rishabh",
  "city": "Bengaluru"
}

Query:

SELECT *
FROM customers
WHERE profile ->> 'city' = 'Bengaluru';

Works just like normal SQL filtering.
```

### Nested JSON

```python
Data:

{
  "name": "Rishabh",
  "address": {
      "city": "Bengaluru",
      "state": "Karnataka"
  }
}

Access city:

SELECT
    profile -> 'address' ->> 'city'
FROM customers;

Result:

Bengaluru
```

### JSON Arrays

```python
Data:

{
  "skills": [
      "SQL",
      "Python",
      "Power BI"
  ]
}

Access:

SELECT
    profile -> 'skills'
FROM customers;

Result:

["SQL","Python","Power BI"]
```

### Updating JSONB

```python
UPDATE customers
SET profile =
jsonb_set(
    profile,
    '{city}',
    '"Mumbai"'
);
```
