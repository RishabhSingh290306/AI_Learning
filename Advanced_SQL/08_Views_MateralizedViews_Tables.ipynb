## 1. Tables

You already know tables.

CREATE TABLE customers (
    customer_id INT,
    customer_name TEXT
);

A table:

✅ Stores data physically

✅ Takes storage space

✅ Can be inserted/updated/deleted

Example:
| customer_id | customer_name |
| ----------- | ------------- |
| 101         | John          |
| 102         | Alice         |

This data is actually stored on disk.


## 2. Views

```python
Think of a View as:

A saved query.

Suppose analysts frequently need:

SELECT
    customer_id,
    UPPER(TRIM(customer_name)) AS clean_name
FROM customers;

Instead of rewriting it daily:

CREATE VIEW clean_customers AS
SELECT
    customer_id,
    UPPER(TRIM(customer_name)) AS clean_name
FROM customers;

Now:

SELECT *
FROM clean_customers;

works like a table.
```

```python
Important

A View stores:

The SQL query

not:

The query result
Example

Table:
| customer_id | customer_name |
| ----------- | ------------- |
| 101         | John          |

View:

SELECT *
FROM clean_customers;

Result:
| customer_id | clean_name |
| ----------- | ---------- |
| 101         | JOHN       |

If you insert:

INSERT INTO customers
VALUES (102, 'Alice');

the view automatically reflects it.

```

## 3. Materialized Views

```python
This is where PostgreSQL gets interesting.

A Materialized View stores:

The query result

not the query alone.

Suppose this query takes 5 minutes:

SELECT
    customer_id,
    SUM(amount)
FROM orders
GROUP BY customer_id;

You don't want to run it every time.

So:

CREATE MATERIALIZED VIEW customer_spending AS
SELECT
    customer_id,
    SUM(amount) AS total_spent
FROM orders
GROUP BY customer_id;

PostgreSQL computes it once and stores the result.
```

```python
Difference
View
Stores SQL

Every query:

Runs again
Materialized View
Stores result

Every query:

Reads stored data

The Catch

Suppose:

Materialized View contains:
| customer_id | total_spent |
| ----------- | ----------- |
| 101         | 5000        |

Now a new order arrives.

Orders table changes.

Materialized View does NOT automatically update.

You must:

REFRESH MATERIALIZED VIEW customer_spending;
```

```python
What You Should Remember

| Feature       | Table | View    | Materialized View |
| ------------- | ----- | ------- | ----------------- |
| Stores Data   | ✅     | ❌       | ✅                 |
| Stores Query  | ❌     | ✅       | ✅                 |
| Fast Reads    | ✅     | Depends | ✅✅                |
| Auto Updates  | ✅     | ✅       | ❌                 |
| Needs Refresh | ❌     | ❌       | ✅                 |
```
