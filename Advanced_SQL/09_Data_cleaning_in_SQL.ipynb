# Data Cleaning in PostgreSQL

### 1. Coalesce()


```python
This is one of the most important functions you'll ever learn.

Suppose:
| customer_id | phone      |
| ----------- | ---------- |
| 101         | 9876543210 |
| 102         | NULL       |
| 103         | 9999999999 |

Query:

SELECT
    customer_id,
    COALESCE(phone, 'Not Provided') AS phone
FROM customers;

Result:
| customer_id | phone        |
| ----------- | ------------ |
| 101         | 9876543210   |
| 102         | Not Provided |
| 103         | 9999999999   |
```

```python
Mental Model

Think:

COALESCE(value1, value2)

means:

"Give me value1 if it exists, otherwise value2."

More Examples
COALESCE(email, phone)

Meaning:

If email exists → use email
Else → use phone
```

### 2. NullIf()

```python
Opposite idea.

Suppose:
| customer_id | salary |
| ----------- | ------ |
| 101         | 50000  |
| 102         | 0      |
| 103         | 75000  |


But your company stores missing salaries as:

0

instead of:

NULL

Convert them:

SELECT
    customer_id,
    NULLIF(salary, 0) AS salary
FROM employees;

Result:
| customer_id | salary |
| ----------- | ------ |
| 101         | 50000  |
| 102         | NULL   |
| 103         | 75000  |
```

### 3. Trim()

```python
Very common.

Data:

"  John  "

Query:

SELECT TRIM('  John  ');

Result:

John
Real Example
SELECT
    TRIM(customer_name)
FROM customers;

Removes leading and trailing spaces.
```

### 4. Upper () and Lower ()

```python
Data:

name
Alice
ALICE
alice
aLiCe

You want consistency.

SELECT
    LOWER(name)
FROM customers;

Result:

alice
alice
alice
alice

Or:

SELECT
    UPPER(name)
FROM customers;

Result:

ALICE
ALICE
ALICE
ALICE
```

### 5. Replace ()

```python
Data:

john@gmail.con

Notice typo:

.con

instead of:

.com

Fix it:

SELECT
    REPLACE(email, '.con', '.com')
FROM customers;
```

### 6. Substring()

```python
Extract part of a string.

Email:

john@gmail.com

Query:

SELECT
    SUBSTRING(email FROM 1 FOR 4)
FROM customers;

Result:

john
```
