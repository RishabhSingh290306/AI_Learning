# Moving Averages

Imagine monthly revenue:
| Month | Revenue |
| ----- | ------- |
| Jan   | 100     |
| Feb   | 500     |
| Mar   | 50      |
| Apr   | 600     |
| May   | 200     |

Notice how noisy it is.

Management asks:

"Can you show the trend instead of the month-to-month fluctuations?"

This is exactly why moving averages exist.

```python
3-Month Moving Average

For each month:

Current Month
+
Previous Month
+
2 Months Ago
----------------
      3
PostgreSQL Query
SELECT
    month,
    revenue,
    AVG(revenue) OVER (
        ORDER BY month
        ROWS BETWEEN 2 PRECEDING
        AND CURRENT ROW
    ) AS moving_avg
FROM sales;
```

```python
Understanding the Window

For January:

Available rows:

Jan

Average:

100

For February:

Available rows:

Jan
Feb

Average:

(100 + 500) / 2
= 300

For March:

Available rows:

Jan
Feb
Mar

Average:

(100 + 500 + 50) / 3
= 216.67

For April:

Window moves forward.

Available rows:

Feb
Mar
Apr

Average:

(500 + 50 + 600) / 3
= 383.33

For May:

Available rows:

Mar
Apr
May

Average:

(50 + 600 + 200) / 3
= 283.33
```

Final Result
| Month | Revenue | Moving Avg |
| ----- | ------- | ---------- |
| Jan   | 100     | 100        |
| Feb   | 500     | 300        |
| Mar   | 50      | 216.67     |
| Apr   | 600     | 383.33     |
| May   | 200     | 283.33     |

Notice how much smoother the trend looks.

```python
Understanding Window Frames

This is the first time we're using a custom frame:

ROWS BETWEEN 2 PRECEDING
AND CURRENT ROW

Read it as:

"Take the current row and the previous 2 rows."

Last 5 Rows
ROWS BETWEEN 4 PRECEDING
AND CURRENT ROW
Last 12 Months
ROWS BETWEEN 11 PRECEDING
AND CURRENT ROW

Very common in business reporting.

Real Data Analyst Use Cases
Revenue Trend
AVG(revenue) OVER (...)
Customer Growth Trend
AVG(new_customers) OVER (...)
Website Traffic Trend
AVG(visitors) OVER (...)
Stock Prices

Moving averages are everywhere in finance.
```
